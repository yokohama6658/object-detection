```python
# ============================================
# YOLO Training Script (Local / Colab Selectable)
# ============================================

!pip install ultralytics   # ← 在 Colab 需要時自行取消註解
from ultralytics import YOLO
import os

# ================================
# 問使用者：你要「本機(local)」還是「Colab + Google Drive」？
# ================================
def select_mode():
  print("請選擇程式執行模式：")
  print("1. 本機(Local)")
  print("2. Colab + Google Drive")
  mode = input("輸入選項 (1 或 2)： ").strip()
  return mode


# =================================
# 模式 A：本機環境
# =================================
def run_local():
  print("你選擇：本機 (Local) 模式")

  # 本機的資料根目錄（自行修改成你的資料夾）
  ROOT = "./"

  data_path = os.path.join(ROOT, "data.yaml")
  output_project_path = os.path.join(ROOT, "runs")
  best_path = os.path.join(ROOT, "runs", "custom_train_exp", "weights", "best.pt")

  # 建立資料夾
  os.makedirs(ROOT, exist_ok=True)
  os.makedirs(output_project_path, exist_ok=True)

  # 檢查 data.yaml
  if not os.path.exists(data_path):
    print(f"找不到 data.yaml：{data_path}")
    print("請確認 data.yaml 已放在本機的 hw5 資料夾內。")
    return

  print("使用本機路徑：", ROOT)

  # 問使用者是否要 resume 或重新訓練
  if os.path.exists(best_path):
    print("偵測到本機已有 best.pt")
    print("1. 繼續訓練 (resume)")
    print("2. 重新訓練 (new)")
    choice = input("選擇 1 或 2： ").strip()

    if choice == "1":
      model = YOLO(best_path)
      print("使用 best.pt 繼續訓練")
      model.train(resume=True)
    else:
      model = YOLO("yolov8n.pt")
      print("使用 yolov8n.pt 重新訓練")
  else:
    model = YOLO("yolov8n.pt")
    print("找不到 best.pt，改為新訓練")

  # 執行訓練
  model.train(
    data=data_path,
    epochs=50,
    imgsz=640,
    project=output_project_path,
    name="custom_train_exp",
    exist_ok=True,
  )

  print("本機訓練完成！結果位於：", output_project_path)


# =================================
# 模式 B：Colab + Google Drive
# =================================
def run_colab():
  print("你選擇：Colab + Drive 模式")

  from google.colab import drive
  drive.mount('/content/drive')

  DRIVE_ROOT = "/content/drive/MyDrive/LEGO"
  data_path = os.path.join(DRIVE_ROOT, "data.yaml")
  output_project_path = os.path.join(DRIVE_ROOT, "runs")
  best_path = os.path.join(DRIVE_ROOT, "runs", "custom_train_exp", "weights", "best.pt")

  os.makedirs(DRIVE_ROOT, exist_ok=True)
  os.makedirs(output_project_path, exist_ok=True)

  if not os.path.exists(data_path):
    print(f"找不到 data.yaml：{data_path}")
    return

  # 問使用者 resume 或新訓練
  if os.path.exists(best_path):
    print("偵測到 Drive 有 best.pt")
    print("1. 繼續訓練 (resume)")
    print("2. 重新訓練 (new)")
    choice = input("選擇 1 或 2： ").strip()

    if choice == "1":
      model = YOLO(best_path)
      print("使用 best.pt 繼續訓練")
      model.train(resume=True)
    else:
      model = YOLO("yolov8n.pt")
      print("使用 yolov8n.pt 重新訓練")
  else:
    model = YOLO("yolov8n.pt")
    print("找不到 best.pt，改為新訓練")

  model.train(
    data=data_path,
    epochs=50,
    imgsz=640,
    project=output_project_path,
    name="custom_train_exp",
    exist_ok=True,
  )

  print("訓練完成！已儲存到 Drive：", output_project_path)


# =======================
# 主程式
# =======================
if __name__ == "__main__":
  mode = select_mode()

  if mode == "1":
    run_local()
  elif mode == "2":
    run_colab()
  else:
    print("錯誤選項，請重新執行程式。")
```
